```python

gamma = 0.9
theta = 0.000001

states = ['S0', 'S1']
V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

iterations = 0

print("Iteration-wise State Values")

while True:
    new_V = V.copy()

    for state in states:
        action_values = {}

        for action, (next_state, reward) in transitions[state].items():
            action_values[action] = reward + gamma * V[next_state]

        new_V[state] = max(action_values.values())

    delta = max(abs(new_V[s] - V[s]) for s in states)
    V = new_V
    iterations += 1

    print(f"Iteration {iterations}: S0 = {V['S0']:.6f}, S1 = {V['S1']:.6f}")

    if delta < theta:
        break

Q = {}
policy = {}

for state in states:
    Q[state] = {}

    for action, (next_state, reward) in transitions[state].items():
        Q[state][action] = reward + gamma * V[next_state]

    policy[state] = max(Q[state], key=Q[state].get)

print("\nFinal Optimal State Values")
for state in states:
    print(f"V*({state}) = {V[state]:.6f}")

print(f"V*(G) = {V['G']:.6f}")

print("\nFinal Action Values")
for state in states:
    for action, value in Q[state].items():
        print(f"Q*({state}, {action}) = {value:.6f}")

print("\nOptimal Policy")
for state in states:
    print(f"Policy({state}) = {policy[state]}")

print("\nNumber of Iterations:", iterations)
```
