```python
import random

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        selected_machine = random.choice(machines)
        action = "Exploration"
    else:
        max_q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        action = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Selected Machine: {selected_machine}")
    print(f"Strategy: {action}")

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    total_reward += reward

    print(f"Updated Q-value of Machine {selected_machine}: "
          f"{Q[selected_machine]:.2f}")

print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nMachine\tN (Count)\tR (Total Reward)\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward received after 10 plays:", total_reward)
```
