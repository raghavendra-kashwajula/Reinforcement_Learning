```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

n = int(input("Enter number of machines: "))
machines = {}

for i in range(n):
    name = chr(65 + i)
    rewards = list(map(int, input(
        f"Enter rewards for Machine {name}: "
    ).split()))

    q = sum(rewards) / len(rewards)
    machines[name] = (rewards, q)

    print(f"\nMachine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")

selected = max(machines, key=lambda x: machines[x][1])

print("\nGreedy Selected Machine:", selected)4
```
