```python

gamma = 0.9
theta = 0.000001

states = ['S0', 'S1']
V = {'S0': 0.0, 'S1': 0.0}

transitions = {
    'S0': {
        'Down': ('S1', -1),
        'Right': ('Wall', -1)
    },
    'S1': {
        'Right': ('G', 100),
        'Up': ('S0', -1)
    }
}

terminal_states = ['Wall', 'G']

iterations = 0

while True:
    delta = 0
    new_V = V.copy()

    for state in states:
        values = []

        for action, (next_state, reward) in transitions[state].items():
            if next_state in terminal_states:
                value = reward
            else:
                value = reward + gamma * V[next_state]

            values.append(value)

        new_V[state] = max(values)
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iterations += 1

    if delta < theta:
        break

policy = {}
Q = {}

for state in states:
    action_values = {}

    for action, (next_state, reward) in transitions[state].items():
        if next_state in terminal_states:
            value = reward
        else:
            value = reward + gamma * V[next_state]

        action_values[action] = value

    best_action = max(action_values, key=action_values.get)
    policy[state] = best_action
    Q[state] = action_values

print("Number of Iterations:", iterations)

print("\nOptimal State Values:")
for state in states:
    print(f"V*({state}) = {V[state]:.6f}")

print("\nAction Values:")
for state in states:
    for action, value in Q[state].items():
        print(f"Q*({state}, {action}) = {value:.6f}")

print("\nOptimal Policy:")
for state in states:
    print(f"Policy({state}) = {policy[state]}")
```
