```python
n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward at time step {i + 1}: "))
    rewards.append(reward)

undiscounted_return = sum(rewards)

gamma = float(input("Enter discount factor gamma: "))

discounted_return = 0

for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]

print("\nEntered reward values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

```python
%%writefile grid_env_det.py

class GridWorld:
    def __init__(self):
        self.states = ["S0", "S1", "S3", "S4", "G"]
        self.actions = ["Up", "Down", "Left", "Right"]
        self.gamma = 0.9

        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        self.transitions = {
            "S0": {
                "Up": "S0",
                "Down": "S0",
                "Left": "S0",
                "Right": "S1"
            },
            "S1": {
                "Up": "S1",
                "Down": "S3",
                "Left": "S0",
                "Right": "G"
            },
            "S3": {
                "Up": "S1",
                "Down": "S3",
                "Left": "S3",
                "Right": "S4"
            },
            "S4": {
                "Up": "G",
                "Down": "S4",
                "Left": "S3",
                "Right": "S4"
            },
            "G": {
                "Up": "G",
                "Down": "G",
                "Left": "G",
                "Right": "G"
            }
        }

    def get_next_state(self, state, action):
        return self.transitions[state][action]

    def get_reward(self, state, action):
        next_state = self.get_next_state(state, action)

        if next_state == "G":
            return self.goal_reward

        return self.normal_reward
```

```python
from grid_env_det import GridWorld

env = GridWorld()

print("2 x 3 GRID WORLD MDP")
print("--------------------")

print("State Space:", env.states)
print("Action Space:", env.actions)
print("Discount Factor:", env.gamma)

print("\nGrid World:")
print("S0  S1  G")
print("X   S3  S4")

state = input("\nEnter current state: ")
action = input("Enter action (Up/Down/Left/Right): ")

if state not in env.states:
    print("Invalid state")
elif action not in env.actions:
    print("Invalid action")
else:
    next_state = env.get_next_state(state, action)
    reward = env.get_reward(state, action)

    print("\nCurrent State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
```
