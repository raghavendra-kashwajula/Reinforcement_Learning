```python
import random

machines = ['A', 'B', 'C']
N_play = 10
epsilon = 0.2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

total_reward = 0

for play in range(N_play):
    r = random.random()
    best_machine = max(Q, key=Q.get)

    if r < epsilon:
        choices = [m for m in machines if m != best_machine]
        selected = random.choice(choices)
    else:
        selected = best_machine

    print(f"\nPlay {play + 1}: Machine {selected}")

    reward = reward_patterns[selected][N[selected] % len(reward_patterns[selected])]
    print("Reward:", reward)

    N[selected] += 1
    R[selected].append(reward)
    Q[selected] += (reward - Q[selected]) / N[selected]
    total_reward += reward

    print("N:", N)
    print("Q:", {m: round(Q[m], 2) for m in machines})

print("\nFinal Results")
print("Machine\tN\tRewards\t\tQ-value")

for m in machines:
    print(f"{m}\t{N[m]}\t{R[m]}\t{Q[m]:.2f}")

print("\nTotal Reward:", total_reward)
```
