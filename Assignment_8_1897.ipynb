```python

grid = {
    'S0': (0, 0),
    'S1': (0, 1),
    'G': (0, 2),
    'X': (1, 0),
    'S3': (1, 1),
    'S4': (1, 2)
}

positions = {v: k for k, v in grid.items()}

actions = {
    'Up': (-1, 0),
    'Down': (1, 0),
    'Left': (0, -1),
    'Right': (0, 1)
}

policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up'
}

gamma = 0.9

def step(state, action):
    row, col = grid[state]
    dr, dc = actions[action]
    new_pos = (row + dr, col + dc)

    if not (0 <= new_pos[0] < 2 and 0 <= new_pos[1] < 3):
        new_pos = (row, col)

    next_state = positions[new_pos]

    if next_state == 'G':
        return next_state, 10

    if next_state == 'X':
        return next_state, -5

    return next_state, -1

def calculate_return(state, first_action=None):
    total = 0
    discount = 1
    first = True
    path = []

    while state not in ['G', 'X']:
        if first and first_action is not None:
            action = first_action
        else:
            action = policy[state]

        next_state, reward = step(state, action)
        path.append((state, action, reward, next_state))

        total += discount * reward
        discount *= gamma
        state = next_state
        first = False

    return total, path

state_value, path = calculate_return('S0')

action_value, action_path = calculate_return('S0', 'Up')

print("State Value Calculation for S0")
for state, action, reward, next_state in path:
    print(f"{state} --{action}--> {next_state}, Reward = {reward}")

print(f"V(S0) = {state_value:.3f}")

print("\nAction Value Calculation for S0 and Up")
for state, action, reward, next_state in action_path:
    print(f"{state} --{action}--> {next_state}, Reward = {reward}")

print(f"Q(S0, Up) = {action_value:.3f}")
```
