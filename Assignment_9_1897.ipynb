```python

gamma = 0.9
theta = 0.000001

V = {'S0': 0.0, 'S1': 0.0}
iterations = 0

while True:
    old_V = V.copy()

    V['S0'] = (
        0.1 * (-1 + gamma * 0) +
        0.9 * (-1 + gamma * old_V['S1'])
    )

    V['S1'] = (
        0.9 * (100 + gamma * 0) +
        0.1 * (-1 + gamma * old_V['S0'])
    )

    delta = max(
        abs(V['S0'] - old_V['S0']),
        abs(V['S1'] - old_V['S1'])
    )

    iterations += 1

    if delta < theta:
        break

Q = {
    'Q(S0, Down)': 0.1 * (-1 + gamma * 0) +
                   0.9 * (-1 + gamma * V['S1']),

    'Q(S0, Right)': -1 + gamma * 0,

    'Q(S1, Right)': 0.9 * (100 + gamma * 0) +
                    0.1 * (-1 + gamma * V['S0']),

    'Q(S1, Up)': -1 + gamma * V['S0']
}

print("Number of Iterations:", iterations)

print("\nState Values:")
for state in V:
    print(f"V({state}) = {V[state]:.6f}")

print("\nAction Values:")
for action, value in Q.items():
    print(f"{action} = {value:.6f}")

print("\nPolicy Consistency Checks:")
print("V(S0) = Q(S0, Down):", abs(V['S0'] - Q['Q(S0, Down)']) < theta)
print("V(S1) = Q(S1, Right):", abs(V['S1'] - Q['Q(S1, Right)']) < theta)
```
