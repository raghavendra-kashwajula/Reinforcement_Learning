```python
# Practical 11: Policy Iteration in a Deterministic Grid World

# Parameters
gamma = 0.9
theta = 1e-6
states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]
terminal_state = "G"

# Deterministic transition function
def transition(state, action):
    if state == "G":
        return "G", 0

    if state == "S0":
        transitions = {
            "Up": ("S0", -1),     # Boundary
            "Down": ("S1", -1),   # Normal movement
            "Left": ("S0", -1),   # Boundary
            "Right": ("S0", -1)   # Wall
        }
    elif state == "S1":
        transitions = {
            "Up": ("S0", -1),     # Normal movement
            "Down": ("S1", -1),   # Boundary
            "Left": ("S1", -1),   # Boundary
            "Right": ("G", 100)   # Goal
        }

    return transitions[action]


# Initial policy and state values
policy = {"S0": "Up", "S1": "Up"}
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

print("Initial Policy:", policy)
print("Initial Values:", V)


# Policy Iteration
iteration = 0

while True:
    iteration += 1
    print(f"\\n========== Policy Iteration {iteration} ==========")

    # Step 1: Policy Evaluation
    while True:
        delta = 0.0

        for state in ["S0", "S1"]:
            old_value = V[state]
            next_state, reward = transition(state, policy[state])
            V[state] = reward + gamma * V[next_state]
            delta = max(delta, abs(old_value - V[state]))

        if delta < theta:
            break

    print("Values after Policy Evaluation:", V)

    # Step 2: Calculate Q-values and improve policy
    Q = {}
    new_policy = {}

    for state in ["S0", "S1"]:
        Q[state] = {}
        for action in actions:
            next_state, reward = transition(state, action)
            Q[state][action] = reward + gamma * V[next_state]

        best_action = max(Q[state], key=Q[state].get)
        new_policy[state] = best_action

    print("Action Values:")
    for state in ["S0", "S1"]:
        print(state, Q[state])

    print("Improved Policy:", new_policy)

    # Step 3: Check policy stability
    if new_policy == policy:
        print("Policy is stable. Algorithm converged.")
        policy = new_policy
        break

    policy = new_policy


# Final Results
print("\\n========== FINAL RESULTS ==========")

print("\\nOptimal State Values:")
for state in states:
    print(f"V*({state}) = {V[state]:.2f}")

print("\\nOptimal Policy:")
for state in ["S0", "S1"]:
    print(f"Policy({state}) = {policy[state]}")

print("\\nFinal Q-value Table:")
print(f"{'State':<8}", end="")
for action in actions:
    print(f"{action:<12}", end="")
print()

for state in ["S0", "S1"]:
    print(f"{state:<8}", end="")
    for action in actions:
        print(f"{Q[state][action]:<12.2f}", end="")
    print()

# Bellman Optimality Verification
print("\\nBellman Optimality Verification:")
for state in ["S0", "S1"]:
    max_q = max(Q[state].values())
    print(f"{state}: V* = {V[state]:.2f}, max Q* = {max_q:.2f}")
```
